## 1. CORE ALGORITHMIC MODULES


```python
def segment_roi_with_qc(img, block_size=16, std_thresh=0.08, min_coherence=0.12, min_roi_ratio=0.08):
    """
    1. Phan vung tien canh (Foreground Mask / ROI) bang Local Variance thich ung.
    2. Bo kiem dinh chat luong (QC Gate) ngan chan anh nhieu ngau nhien khong phai sinh trac:
       - Yeu cau do ket dinh huong dong chay (Orientation Coherence) >= min_coherence (mac dinh 0.12).
       - Yeu cau ti le dien tich vung van tay hop le >= min_roi_ratio (mac dinh 8%).
    """
    h, w = img.shape
    norm_float = img.astype(np.float32) / 255.0

    kernel = np.ones((block_size, block_size), dtype=np.float32) / (block_size * block_size)
    local_mean = cv2.filter2D(norm_float, -1, kernel)
    local_sq_mean = cv2.filter2D(norm_float**2, -1, kernel)
    local_var = np.maximum(local_sq_mean - local_mean**2, 0.0)
    local_std = np.sqrt(local_var)

    mask = (local_std > std_thresh).astype(np.uint8) * 255

    close_kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (15, 15))
    mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, close_kernel)

    erode_kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (9, 9))
    mask_eroded = cv2.morphologyEx(mask, cv2.MORPH_ERODE, erode_kernel)

    # Tinh Orientation Coherence kiem tra tinh dong nhat cua cac dai van
    gx = cv2.Sobel(img.astype(np.float32), cv2.CV_32F, 1, 0, ksize=3)
    gy = cv2.Sobel(img.astype(np.float32), cv2.CV_32F, 0, 1, ksize=3)
    vx = 2.0 * gx * gy
    vy = gx**2 - gy**2
    ksize_smooth = block_size * 2 + 1
    vx_smooth = cv2.GaussianBlur(vx, (ksize_smooth, ksize_smooth), 0)
    vy_smooth = cv2.GaussianBlur(vy, (ksize_smooth, ksize_smooth), 0)
    energy_smooth = cv2.GaussianBlur(gx**2 + gy**2, (ksize_smooth, ksize_smooth), 0)

    coherence = np.sqrt(vx_smooth**2 + vy_smooth**2) / (energy_smooth + 1e-5)
    roi_pixels = mask > 0
    roi_area_ratio = np.sum(roi_pixels) / float(h * w)

    if np.sum(roi_pixels) > 0:
        mean_coherence = float(np.mean(coherence[roi_pixels]))
    else:
        mean_coherence = 0.0

    is_valid_biometric = (mean_coherence >= min_coherence) and (roi_area_ratio >= min_roi_ratio)
    qc_status = "PASS (Biometric QC Validated)" if is_valid_biometric else "REJECT (QC Reject - Low Coherence / Flat Contrast)"

    qc_info = {
        "is_valid": is_valid_biometric,
        "mean_coherence": mean_coherence,
        "roi_ratio": roi_area_ratio * 100.0,
        "status_text": qc_status
    }
    return mask, mask_eroded, qc_info


def normalize_roi(img, mask, m0=100.0, v0=100.0):
    """
    Chuan hoa anh muc xam chi ben trong vung van tay (ROI):
    Bao dam do sang va do tuong phan dong deu ma khong bi lech boi nen trang ngoai vien.
    """
    norm = img.astype(np.float32)
    roi_pixels = norm[mask > 0]
    if len(roi_pixels) == 0:
        return img

    m = np.mean(roi_pixels)
    v = np.var(roi_pixels)
    if v < 1e-5:
        v = 1e-5

    coeff = np.sqrt(v0 * ((norm - m)**2) / v)
    result = np.where(norm > m, m0 + coeff, m0 - coeff)
    result = np.clip(result, 0, 255).astype(np.uint8)
    result[mask == 0] = 255
    return result


def estimate_orientation_ridge(img, mask, block_size=16):
    """
    Uoc luong Truong huong Dong van (Orientation Field) thich ung:
    - Su dung bieu dien goc kep: vx = 2 * Gx * Gy, vy = Gx^2 - Gy^2.
    - theta_grad = 0.5 * atan2(vx, vy) la huong Gradient phap tuyen.
    - theta_ridge = (theta_grad + pi/2) mod pi la tiep tuyen dong van.
    - Adaptive Smoothing: Giam do rong lam muot tai vung co do bien thien huong lon de bao toan tam xoay Core/Delta.
    """
    h, w = img.shape
    gx = cv2.Sobel(img.astype(np.float32), cv2.CV_32F, 1, 0, ksize=3)
    gy = cv2.Sobel(img.astype(np.float32), cv2.CV_32F, 0, 1, ksize=3)

    vx = 2.0 * gx * gy
    vy = gx**2 - gy**2

    # Lam muot 2 lop: lop nho (17x17) bao ve Core/Delta, lop lon (33x33) cho dong van thang
    vx_fine = cv2.GaussianBlur(vx, (17, 17), 0)
    vy_fine = cv2.GaussianBlur(vy, (17, 17), 0)
    vx_coarse = cv2.GaussianBlur(vx, (33, 33), 0)
    vy_coarse = cv2.GaussianBlur(vy, (33, 33), 0)

    # Do bien thien goc cuc bo
    grad_div = np.abs(cv2.Sobel(vx_fine, cv2.CV_32F, 1, 0)) + np.abs(cv2.Sobel(vy_fine, cv2.CV_32F, 0, 1))
    weight_fine = np.clip(grad_div / (np.percentile(grad_div, 95) + 1e-5), 0.0, 1.0)

    vx_smooth = weight_fine * vx_fine + (1.0 - weight_fine) * vx_coarse
    vy_smooth = weight_fine * vy_fine + (1.0 - weight_fine) * vy_coarse

    theta_grad = 0.5 * np.arctan2(vx_smooth, vy_smooth)
    theta_ridge = (theta_grad + np.pi / 2.0) % np.pi
    theta_ridge[mask == 0] = 0.0
    return theta_ridge, theta_grad


def estimate_ridge_wavelength(img, mask, default_lambd=10.0):
    """
    Uoc luong buoc song dong cua cac dai van (Dynamic Ridge Wavelength Estimation):
    Khac phuc lo hong gan cung lambda = 10.0 px khi anh co do phan giai khac hoac ngon tay nho/lon.
    """
    h, w = img.shape
    cy, cx = h // 2, w // 2
    wavelength_candidates = []

    for dy in [-40, -20, 0, 20, 40]:
        y = cy + dy
        if 0 <= y < h:
            line = img[y, :].astype(np.float32)
            m_line = mask[y, :] > 0
            if np.sum(m_line) > 50:
                seg = line[m_line]
                seg = seg - np.mean(seg)
                fft = np.abs(np.fft.rfft(seg))
                freqs = np.fft.rfftfreq(len(seg))
                if len(fft) > 3:
                    fft[0] = 0.0 # Triet tieu thanh phan DC
                    peak_idx = int(np.argmax(fft))
                    if freqs[peak_idx] > 0:
                        lambd_val = 1.0 / freqs[peak_idx]
                        if 6.0 <= lambd_val <= 18.0:
                            wavelength_candidates.append(lambd_val)

    if len(wavelength_candidates) >= 2:
        return float(np.median(wavelength_candidates))
    return float(default_lambd)


def apply_gabor_filter_bank(img, orientation_ridge, mask, num_orientations=16, ksize=17, sigma=3.5, lambd=None):
    """
    Ngan hang Bo loc Gabor 2D toan cuc (16 huong roi rac):
    - Tu dong su dung buoc song dong duoc uoc luong neu lambd=None.
    - Gabor dao dong vuong goc tiep tuyen dong van (ang + pi/2).
    - Triet tieu hoan toan duong noi khoi (Block artifacts).
    """
    h, w = img.shape
    img_float = img.astype(np.float32)

    if lambd is None:
        lambd = estimate_ridge_wavelength(img, mask, default_lambd=10.0)

    filtered_bank = np.zeros((num_orientations, h, w), dtype=np.float32)
    angles = np.linspace(0, np.pi, num_orientations, endpoint=False)

    for k, ang in enumerate(angles):
        kernel = cv2.getGaborKernel((ksize, ksize), sigma=sigma, theta=ang + np.pi/2.0,
                                    lambd=lambd, gamma=0.5, psi=0, ktype=cv2.CV_32F)
        filtered_bank[k] = cv2.filter2D(img_float, -1, kernel)

    ori_wrapped = orientation_ridge % np.pi
    angle_idx = np.round(ori_wrapped / (np.pi / num_orientations)).astype(np.int32) % num_orientations
    enhanced = np.take_along_axis(filtered_bank, angle_idx[np.newaxis, :, :], axis=0)[0]

    enhanced_roi = enhanced[mask > 0]
    if len(enhanced_roi) > 0:
        min_v = np.percentile(enhanced_roi, 1)
        max_v = np.percentile(enhanced_roi, 99)
        enhanced_clipped = np.clip(enhanced, min_v, max_v)
        enhanced_norm = ((enhanced_clipped - min_v) / max(max_v - min_v, 1e-4) * 255.0).astype(np.uint8)
    else:
        enhanced_norm = img.copy()

    enhanced_norm[mask == 0] = 255
    return enhanced_norm


def fast_thinning_zhang_suen(binary_img, mask):
    """Ep xuong Zhang-Suen toi uu hoa toc do cao bang NumPy Vectorized"""
    if hasattr(cv2, 'ximgproc') and hasattr(cv2.ximgproc, 'thinning'):
        skel = cv2.ximgproc.thinning(binary_img, thinningType=cv2.ximgproc.THINNING_ZHANGSUEN)
        skel[mask == 0] = 0
        return skel

    im = (binary_img > 0).astype(np.uint8)
    h, w = im.shape

    while True:
        p2 = np.roll(im, -1, axis=0)
        p3 = np.roll(np.roll(im, -1, axis=0), 1, axis=1)
        p4 = np.roll(im, 1, axis=1)
        p5 = np.roll(np.roll(im, 1, axis=0), 1, axis=1)
        p6 = np.roll(im, 1, axis=0)
        p7 = np.roll(np.roll(im, 1, axis=0), -1, axis=1)
        p8 = np.roll(im, -1, axis=1)
        p9 = np.roll(np.roll(im, -1, axis=0), -1, axis=1)

        b = p2 + p3 + p4 + p5 + p6 + p7 + p8 + p9
        pattern = [p2, p3, p4, p5, p6, p7, p8, p9, p2]
        a = sum((pattern[i] == 0) & (pattern[i+1] == 1) for i in range(8))

        del1 = (im == 1) & (b >= 2) & (b <= 6) & (a == 1) & (p2 * p4 * p6 == 0) & (p4 * p6 * p8 == 0)
        im[del1] = 0

        p2 = np.roll(im, -1, axis=0)
        p3 = np.roll(np.roll(im, -1, axis=0), 1, axis=1)
        p4 = np.roll(im, 1, axis=1)
        p5 = np.roll(np.roll(im, 1, axis=0), 1, axis=1)
        p6 = np.roll(im, 1, axis=0)
        p7 = np.roll(np.roll(im, 1, axis=0), -1, axis=1)
        p8 = np.roll(im, -1, axis=1)
        p9 = np.roll(np.roll(im, -1, axis=0), -1, axis=1)
        b = p2 + p3 + p4 + p5 + p6 + p7 + p8 + p9
        pattern = [p2, p3, p4, p5, p6, p7, p8, p9, p2]
        a = sum((pattern[i] == 0) & (pattern[i+1] == 1) for i in range(8))

        del2 = (im == 1) & (b >= 2) & (b <= 6) & (a == 1) & (p2 * p4 * p8 == 0) & (p2 * p6 * p8 == 0)
        im[del2] = 0

        if not np.any(del1) and not np.any(del2):
            break

    skel = (im * 255).astype(np.uint8)
    skel[mask == 0] = 0
    return skel


def smart_prune_spurs(skeleton, max_spur_length=6):
    """
    Lan theo tung dau mut (Endpoint). Chi xoa neu nhanh gai ngan (<= max_spur_length)
    va dau mut gan vao mot diem re nhanh (Bifurcation).
    Bao toan tuyet doi cac dau mut dong van that khong dinh vao nga ba.
    """
    skel = (skeleton > 0).astype(np.uint8)
    h, w = skel.shape

    for _ in range(2):
        pad = np.pad(skel, 1, mode='constant', constant_values=0)
        neighbors = (
            pad[0:h, 0:w] + pad[0:h, 1:w+1] + pad[0:h, 2:w+2] +
            pad[1:h+1, 0:w]                 + pad[1:h+1, 2:w+2] +
            pad[2:h+2, 0:w] + pad[2:h+2, 1:w+1] + pad[2:h+2, 2:w+2]
        ) * skel

        endpoints = np.argwhere((neighbors == 1) & (skel == 1))
        to_remove = set()

        for er, ec in endpoints:
            path = [(er, ec)]
            curr_r, curr_c = er, ec
            prev_r, prev_c = -1, -1
            is_spur = False

            for _step in range(max_spur_length):
                nb_list = []
                for dr in [-1, 0, 1]:
                    for dc in [-1, 0, 1]:
                        if dr == 0 and dc == 0: continue
                        nr, nc = curr_r + dr, curr_c + dc
                        if 0 <= nr < h and 0 <= nc < w and skel[nr, nc] == 1:
                            if (nr, nc) != (prev_r, prev_c):
                                nb_list.append((nr, nc))

                if len(nb_list) == 0:
                    break
                elif len(nb_list) == 1:
                    nxt_r, nxt_c = nb_list[0]
                    path.append((nxt_r, nxt_c))
                    prev_r, prev_c = curr_r, curr_c
                    curr_r, curr_c = nxt_r, nxt_c
                else:
                    is_spur = True
                    break

            if is_spur and len(path) <= max_spur_length:
                for pr, pc in path:
                    to_remove.add((pr, pc))

        if not to_remove:
            break
        for r, c in to_remove:
            skel[r, c] = 0

    return (skel * 255).astype(np.uint8)


def extract_validated_minutiae(skeleton, orientation_ridge, mask_eroded, min_dist=10, **kwargs):
    if 'min_distance' in kwargs and kwargs['min_distance'] is not None:
        min_dist = kwargs['min_distance']
    """
    Trich xuat Minutiae co kiem dinh chat luong:
    - Loc ung vien nam ngoai ROI da co.
    - Crossing Number (CN): CN=1 (Ending), CN=3 (Bifurcation).
    - Luu tru bo tu thong tin: (x, y, type, theta).
    - Khử cap nut gay gia do seo va diem re nhanh nam sat dau mut.
    """
    h, w = skeleton.shape
    skel_norm = (skeleton > 0).astype(np.uint8)

    raw_endings = []
    raw_bifurcations = []

    for r in range(1, h - 1):
        for c in range(1, w - 1):
            if skel_norm[r, c] == 1 and mask_eroded[r, c] > 0:
                p = [
                    skel_norm[r-1, c],   skel_norm[r-1, c+1], skel_norm[r, c+1],
                    skel_norm[r+1, c+1], skel_norm[r+1, c],   skel_norm[r+1, c-1],
                    skel_norm[r, c-1],   skel_norm[r-1, c-1], skel_norm[r-1, c]
                ]
                cn = 0.5 * sum(abs(int(p[i]) - int(p[i+1])) for i in range(8))
                th = orientation_ridge[r, c]

                if cn == 1:
                    raw_endings.append({"x": c, "y": r, "type": "ending", "theta": th})
                elif cn == 3:
                    raw_bifurcations.append({"x": c, "y": r, "type": "bifurcation", "theta": th})

    valid_endings = []
    valid_bifurcations = []

    discard_e = set()
    for i in range(len(raw_endings)):
        if i in discard_e: continue
        e1 = raw_endings[i]
        for j in range(i + 1, len(raw_endings)):
            if j in discard_e: continue
            e2 = raw_endings[j]
            d = math.hypot(e1["x"] - e2["x"], e1["y"] - e2["y"])
            if d < min_dist:
                discard_e.add(i)
                discard_e.add(j)
                break
        if i not in discard_e:
            valid_endings.append(e1)

    for b in raw_bifurcations:
        is_spurious = False
        for e in valid_endings:
            if math.hypot(b["x"] - e["x"], b["y"] - e["y"]) < min_dist:
                is_spurious = True
                break
        if not is_spurious:
            valid_bifurcations.append(b)

    return valid_endings, valid_bifurcations


def visualize_orientation_field(img, orientation_ridge, mask, step=16):
    """Truc quan hoa truong huong tiep tuyen dong van dang kim vang High-Tech"""
    h, w = img.shape
    vis = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)
    length = step // 2 - 1

    for r in range(step // 2, h, step):
        for c in range(step // 2, w, step):
            if mask[r, c] > 0:
                th = orientation_ridge[r, c]
                dx = int(length * math.cos(th))
                dy = int(length * math.sin(th))
                cv2.line(vis, (c - dx, r - dy), (c + dx, r + dy), (0, 220, 255), 1, cv2.LINE_AA)
    return vis


def visualize_minutiae_map(skeleton, endings, bifurcations):
    """Truc quan hoa ban do Minutiae (Do: Ending, Xanh neon: Bifurcation)"""
    vis = cv2.cvtColor(skeleton, cv2.COLOR_GRAY2BGR)
    for e in endings:
        pt = (e["x"], e["y"]) if isinstance(e, dict) else (int(e[0]), int(e[1]))
        cv2.circle(vis, pt, 3, (0, 0, 255), -1)
    for b in bifurcations:
        pt = (b["x"], b["y"]) if isinstance(b, dict) else (int(b[0]), int(b[1]))
        cv2.circle(vis, pt, 3, (0, 255, 128), -1)
    return vis
```

## 2. DỮ LIỆU MÔ PHỎNG & SO KHỚP CÓ ĐĂNG KÝ (ALIGNMENT & MATCHING)


```python
def generate_synthetic_fingerprint_advanced(pattern_type="whorl", seed=42, condition="normal", noise_level=0.06, rot_deg=0.0, shift=(0, 0)):
    """
    Sinh anh van tay toan hoc kiem thu tieu chuan:
    - condition=normal: Van tay ro net tieu chuan
    - condition=broken_scars: Van tay bi seo/vet cat ngang
    - condition=faint_dry: Van tay kho, do tuong phan thap
    - condition=wet_smudged: Van tay uot nhoet muc
    """
    np.random.seed(seed)
    h, w = 320, 240
    y, x = np.ogrid[:h, :w]
    freq = 2.0 * np.pi / 10.0

    if pattern_type == "whorl":
        cy, cx = h // 2 + shift[1], w // 2 + shift[0]
        r = np.sqrt((x - cx)**2 + 1.2 * (y - cy)**2)
        theta = np.arctan2(y - cy, x - cx)
        ridges = np.sin(freq * (r + 14.0 * np.cos(theta * 2)))
        mask = (((x - cx)**2 / (85**2) + (y - cy)**2 / (120**2)) < 1.0)
    elif pattern_type == "loop":
        cy, cx = h // 2 + 25 + shift[1], w // 2 - 20 + shift[0]
        r = np.sqrt((x - cx)**2 + 1.6 * (y - cy)**2)
        theta = np.arctan2(y - cy, x - cx)
        ridges = np.sin(freq * (r + 18.0 * np.sin(theta)))
        mask = (((x - w//2)**2 / (85**2) + (y - h//2)**2 / (120**2)) < 1.0)
    else:
        cy, cx = h // 2 + shift[1], w // 2 + shift[0]
        ridges = np.sin(freq * (y + 35.0 * np.exp(-((x - cx)**2) / 2500.0)))
        mask = (((x - cx)**2 / (85**2) + (y - cy)**2 / (120**2)) < 1.0)

    img = np.zeros((h, w), dtype=np.float32)

    if condition == "faint_dry":
        img[mask] = (ridges[mask] + 1.0) * 35.0 + 155.0
        img[~mask] = 230.0
        patchy_noise = np.random.uniform(0.7, 1.0, (h, w))
        img[mask] = img[mask] * patchy_noise[mask]
        noise = np.random.normal(0, 16.0, (h, w))
    elif condition == "wet_smudged":
        wet_ridges = np.where(ridges > -0.2, 1.0, ridges)
        img[mask] = (wet_ridges[mask] + 1.0) * 60.0 + 20.0
        img[~mask] = 240.0
        noise = np.random.normal(0, 8.0, (h, w))
    else:
        img[mask] = (ridges[mask] + 1.0) * 105.0 + 35.0
        img[~mask] = 235.0
        noise = np.random.normal(0, noise_level * 50.0, (h, w))

    final_img = np.clip(img + noise, 0, 255).astype(np.uint8)

    if condition == "broken_scars":
        cuts = [
            ((int(w * 0.25), int(h * 0.35)), (int(w * 0.75), int(h * 0.45)), 3),
            ((int(w * 0.30), int(h * 0.60)), (int(w * 0.70), int(h * 0.52)), 2),
            ((int(w * 0.45), int(h * 0.25)), (int(w * 0.55), int(h * 0.75)), 2)
        ]
        for pt1, pt2, thick in cuts:
            cv2.line(final_img, pt1, pt2, 230, thick)
        drop_mask = np.random.rand(h, w) < 0.05
        final_img[drop_mask & mask] = 220

    if abs(rot_deg) > 0.01:
        M = cv2.getRotationMatrix2D((w // 2, h // 2), rot_deg, 1.0)
        final_img = cv2.warpAffine(final_img, M, (w, h), borderValue=235)

    return final_img


def match_minutiae_rigid_alignment(pts1, pts2, max_dist=14, max_angle_diff=np.pi/6, return_info=False):
    """
    Dang ky Can chinh Toan cuc Bat bien Goc quay 360 do (Full 360 Rotation Invariant Rigid Alignment)
    ket hop Dung sai Bien dang Dan hoi phi tuyen (Non-linear Elastic Distortion):
    1. Lay mau Minutiae phan bo da goc phan tu (4 quadrants) de bao quat toan ngon tay.
    2. Hough Voting da thang do (Coarse-to-Fine 3D Hough Voting):
       - Quet toan dai 36 goc [-180, +180] (buoc 10 do).
       - Bo phieu tich luy Accumulator mang O(N1*N2), toc do cuc nhanh (< 50ms).
       - Tinh chinh goc xoay quanh dinh cuc dai trong pham vi +-8 do (buoc 2 do).
    3. Dung sai ban kinh dan hoi phi tuyen: r(d) = max_dist + 0.04 * d_center.
    4. Doi sanh Minutiae luong phan tham lam (Greedy Bipartite Match) voi rang buoc kep.
    """
    if len(pts1) == 0 or len(pts2) == 0:
        info = {"rot_deg": 0.0, "best_rot_rad": 0.0, "dx": 0.0, "dy": 0.0, "votes": 0}
        if return_info:
            return 0.0, [], info
        return 0.0, []

    c1x = float(np.mean([p["x"] for p in pts1]))
    c1y = float(np.mean([p["y"] for p in pts1]))

    def sample_distributed_candidates(points, max_count=48):
        if len(points) <= max_count:
            return points
        sorted_by_dist = sorted(points, key=lambda p: math.hypot(p["x"] - c1x, p["y"] - c1y))
        step = max(1, len(points) // max_count)
        return sorted_by_dist[::step][:max_count]

    candidates_1 = sample_distributed_candidates(pts1, max_count=40)
    candidates_2 = sample_distributed_candidates(pts2, max_count=40)

    # 1. COARSE SEARCH: Quet 36 goc [-pi, +pi) (khoang cach 10 do = pi/18 rad)
    num_coarse = 36
    coarse_rot_angles = [i * (2.0 * math.pi / num_coarse) - math.pi for i in range(num_coarse)]

    best_votes = -1
    best_rot = 0.0
    best_dx = 0.0
    best_dy = 0.0
    bin_size = max(10.0, float(max_dist))

    for rot in coarse_rot_angles:
        cos_r = math.cos(rot)
        sin_r = math.sin(rot)
        acc = {}
        for p1 in candidates_1:
            rx = cos_r * (p1["x"] - c1x) - sin_r * (p1["y"] - c1y) + c1x
            ry = sin_r * (p1["x"] - c1x) + cos_r * (p1["y"] - c1y) + c1y
            rtheta = (p1["theta"] + rot) % math.pi
            for p2 in candidates_2:
                if p1["type"] != p2["type"]:
                    continue
                adiff = abs(rtheta - p2["theta"])
                adiff = min(adiff, math.pi - adiff)
                if adiff > max_angle_diff:
                    continue
                dx = p2["x"] - rx
                dy = p2["y"] - ry
                bx = int(round(dx / bin_size))
                by = int(round(dy / bin_size))
                key = (bx, by)
                if key not in acc:
                    acc[key] = [0, 0.0, 0.0]
                acc[key][0] += 1
                acc[key][1] += dx
                acc[key][2] += dy

        for (bx, by), val in acc.items():
            votes = val[0]
            if votes > best_votes:
                best_votes = votes
                best_rot = rot
                best_dx = val[1] / votes
                best_dy = val[2] / votes

    # 2. FINE REFINEMENT: Tinh chinh trong khoang +-8 do quanh best_rot (buoc 2 do)
    fine_deg_offsets = range(-8, 9, 2)
    fine_rot_angles = [best_rot + math.radians(deg) for deg in fine_deg_offsets]

    for rot in fine_rot_angles:
        cos_r = math.cos(rot)
        sin_r = math.sin(rot)
        acc = {}
        for p1 in candidates_1:
            rx = cos_r * (p1["x"] - c1x) - sin_r * (p1["y"] - c1y) + c1x
            ry = sin_r * (p1["x"] - c1x) + cos_r * (p1["y"] - c1y) + c1y
            rtheta = (p1["theta"] + rot) % math.pi
            for p2 in candidates_2:
                if p1["type"] != p2["type"]:
                    continue
                adiff = abs(rtheta - p2["theta"])
                adiff = min(adiff, math.pi - adiff)
                if adiff > max_angle_diff:
                    continue
                dx = p2["x"] - rx
                dy = p2["y"] - ry
                bx = int(round(dx / bin_size))
                by = int(round(dy / bin_size))
                key = (bx, by)
                if key not in acc:
                    acc[key] = [0, 0.0, 0.0]
                acc[key][0] += 1
                acc[key][1] += dx
                acc[key][2] += dy

        for (bx, by), val in acc.items():
            votes = val[0]
            if votes >= best_votes:
                best_votes = votes
                best_rot = rot
                best_dx = val[1] / votes
                best_dy = val[2] / votes

    # 3. BIPARTITE MATCHING VOI DUNG SAI DAN HOI
    matched_pairs = []
    used_pts2 = set()
    cos_r = math.cos(best_rot)
    sin_r = math.sin(best_rot)

    for p1 in pts1:
        rx = cos_r * (p1["x"] - c1x) - sin_r * (p1["y"] - c1y) + c1x + best_dx
        ry = sin_r * (p1["x"] - c1x) + cos_r * (p1["y"] - c1y) + c1y + best_dy
        rotated_theta = (p1["theta"] + best_rot) % math.pi

        dist_from_center = math.hypot(p1["x"] - c1x, p1["y"] - c1y)
        dynamic_dist_thresh = max_dist + 0.04 * dist_from_center

        best_d = 999.0
        best_idx2 = -1

        for idx2, p2 in enumerate(pts2):
            if idx2 in used_pts2 or p1["type"] != p2["type"]:
                continue
            angle_diff = abs(rotated_theta - p2["theta"])
            angle_diff = min(angle_diff, np.pi - angle_diff)
            if angle_diff > max_angle_diff:
                continue
            dist = math.hypot(rx - p2["x"], ry - p2["y"])
            if dist < best_d and dist <= dynamic_dist_thresh:
                best_d = dist
                best_idx2 = idx2

        if best_idx2 != -1:
            matched_pairs.append(((int(round(p1["x"])), int(round(p1["y"]))),
                                  (int(round(pts2[best_idx2]["x"])), int(round(pts2[best_idx2]["y"])))))
            used_pts2.add(best_idx2)

    total_possible = (len(pts1) + len(pts2)) / 2.0
    score = (len(matched_pairs) / total_possible) * 100.0 if total_possible > 0 else 0.0

    rot_deg_normalized = math.degrees(best_rot)
    while rot_deg_normalized > 180.0:
        rot_deg_normalized -= 360.0
    while rot_deg_normalized <= -180.0:
        rot_deg_normalized += 360.0

    info = {
        "rot_deg": rot_deg_normalized,
        "best_rot_rad": best_rot,
        "dx": best_dx,
        "dy": best_dy,
        "votes": best_votes
    }

    if return_info:
        return min(score, 100.0), matched_pairs, info
    return min(score, 100.0), matched_pairs
```

```python
def run_pipeline_inspector(input_img, preset_choice):
    """Tab 1: Mo xe 6 cong doan tien xu ly va trich xuat dac trung sinh trac"""
    t0 = time.time()
    choice_str = str(preset_choice)

    # 1. Kiem tra che do va nap anh (Uu tien anh thuc te FVC)
    if "Sử dụng ảnh" in choice_str or "tai len" in choice_str.lower() or "upload" in choice_str.lower():
        if input_img is None:
            gray = cv2.imread("real_fingerprints/01_real_normal_clear.png", cv2.IMREAD_GRAYSCALE)
            if gray is None:
                gray = generate_synthetic_fingerprint_advanced(pattern_type="whorl", seed=42)
        else:
            gray = cv2.cvtColor(input_img, cv2.COLOR_RGB2GRAY) if len(input_img.shape) == 3 else input_img
            gray = cv2.resize(gray, (240, 320))
            clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
            gray = clahe.apply(gray)
    elif "2" in choice_str or "DUT" in choice_str.upper() or "SEO" in choice_str.upper() or "broken" in choice_str.lower():
        gray = cv2.imread("real_fingerprints/02_real_broken_scars.png", cv2.IMREAD_GRAYSCALE)
        if gray is None:
            gray = generate_synthetic_fingerprint_advanced(pattern_type="whorl", seed=42, condition="broken_scars")
    elif "3" in choice_str or "KHO" in choice_str.upper() or "MO" in choice_str.upper() or "dry" in choice_str.lower():
        gray = cv2.imread("real_fingerprints/03_real_faint_dry.png", cv2.IMREAD_GRAYSCALE)
        if gray is None:
            gray = generate_synthetic_fingerprint_advanced(pattern_type="whorl", seed=42, condition="faint_dry")
    elif "4" in choice_str or "UOT" in choice_str.upper() or "BET" in choice_str.upper() or "wet" in choice_str.lower():
        gray = cv2.imread("real_fingerprints/04_real_wet_smudged.png", cv2.IMREAD_GRAYSCALE)
        if gray is None:
            gray = generate_synthetic_fingerprint_advanced(pattern_type="whorl", seed=42, condition="wet_smudged")
    else:  # Mac dinh la Mo phong 1: Chuan ro net
        gray = cv2.imread("real_fingerprints/01_real_normal_clear.png", cv2.IMREAD_GRAYSCALE)
        if gray is None:
            gray = generate_synthetic_fingerprint_advanced(pattern_type="whorl", seed=42, condition="normal")

    if gray is not None:
        gray = cv2.resize(gray, (240, 320))

    s1_raw = cv2.cvtColor(gray, cv2.COLOR_GRAY2RGB)

    mask, mask_eroded, qc_info = segment_roi_with_qc(gray)
    norm = normalize_roi(gray, mask)
    s2_norm = cv2.cvtColor(norm, cv2.COLOR_GRAY2RGB)

    orient_ridge, _ = estimate_orientation_ridge(norm, mask)
    s3_orient = cv2.cvtColor(visualize_orientation_field(norm, orient_ridge, mask), cv2.COLOR_BGR2RGB)

    lambd_est = estimate_ridge_wavelength(norm, mask)
    enhanced = apply_gabor_filter_bank(norm, orient_ridge, mask, lambd=lambd_est)
    s4_gabor = cv2.cvtColor(enhanced, cv2.COLOR_GRAY2RGB)

    _, binary = cv2.threshold(enhanced, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
    binary[mask == 0] = 0
    skel = fast_thinning_zhang_suen(binary, mask)
    skel = smart_prune_spurs(skel, max_spur_length=6)
    s5_skel = cv2.cvtColor(skel, cv2.COLOR_GRAY2RGB)

    if qc_info["is_valid"]:
        endings, bifurcations = extract_validated_minutiae(skel, orient_ridge, mask_eroded)
    else:
        endings, bifurcations = [], []

    vis_min = visualize_minutiae_map(skel, endings, bifurcations)
    s6_minutiae = cv2.cvtColor(vis_min, cv2.COLOR_BGR2RGB)
    total_time_ms = (time.time() - t0) * 1000.0

    metrics_md = f"""
### BÁO CÁO KIỂM CHUẨN ĐẶC TẢ SINH TRẮC HỌC (V5 - REAL ASSETS)
| Tiêu chí Kiểm định Sinh trắc | Kết quả Đo đạc Thực nghiệm | Cơ sở Khoa học & Cơ chế Phòng vệ |
| :--- | :--- | :--- |
| **Cửa kiểm định (QC Gate)** | **`{qc_info['status_text']}`** | Độ kết dính Coherence = `{qc_info['mean_coherence']:.2f}` (Ngưỡng >= 0.12, loại trừ ảnh nhiễu) |
| **Vùng vân hợp lệ (ROI)** | `Tỉ lệ diện tích: {qc_info['roi_ratio']:.1f}%` | Co biên xói mòn hình thái học triệt tiêu hoàn toàn Minutiae giả ở rìa |
| **Bước sóng dải vân** | **`lambda = {lambd_est:.1f} px`** | Ước lượng tần số không gian động qua 1D-FFT, chống bết dính và đứt dải vân |
| **Ngân hàng lọc Gabor** | `16 Hướng không gian` | Lọc dải vân định hướng liên tục, kết nối vết sẹo và đứt nét |
| **Điểm kết thúc (Đỏ)** | **`{len(endings)}` Minutiae** | Smart Spur Pruning bảo toàn đầu mút thật, gọt tỉa dải gai ngắn <= 6px |
| **Điểm rẽ nhánh (Xanh)** | **`{len(bifurcations)}` Minutiae** | Crossing Number CN = 3 sau làm mảnh Zhang-Suen 1px chuẩn quốc tế |
| **Thời gian tiền xử lý** | **`{total_time_ms:.1f} ms`** | Tốc độ đáp ứng thời gian thực cho hệ thống sinh trắc học (< 200ms) |
"""
    return s1_raw, s2_norm, s3_orient, s4_gabor, s5_skel, s6_minutiae, metrics_md


def run_biometric_matching(img1_in, img2_in, scenario_choice, custom_rot_param, max_dist_param, threshold_param):
    """Tab 2: Doi chieu xac thuc 2 dau van tay voi Dang ky Can chinh Rigid Alignment 360 do"""
    t0 = time.time()
    choice_str = str(scenario_choice)

    # 1. Nap anh F1 (Template) va F2 (Query)
    real1 = cv2.imread("real_fingerprints/01_real_normal_clear.png", cv2.IMREAD_GRAYSCALE)
    if real1 is not None:
        real1 = cv2.resize(real1, (240, 320))
    else:
        real1 = generate_synthetic_fingerprint_advanced(pattern_type="whorl", seed=42, noise_level=0.06)

    if "G" in choice_str or "mạo danh" in choice_str.lower() or "impostor" in choice_str.lower():
        # Kich ban G: Hai nguoi hoan toan khac nhau (Impostor)
        real2 = cv2.imread("real_fingerprints/02_real_broken_scars.png", cv2.IMREAD_GRAYSCALE)
        if real2 is not None:
            raw2 = cv2.resize(real2, (240, 320))
        else:
            raw2 = generate_synthetic_fingerprint_advanced(pattern_type="loop", seed=99, noise_level=0.08)
        raw1 = real1
    elif "Tự tải" in choice_str or "tai len" in choice_str.lower():
        if img1_in is None:
            raw1 = real1
        else:
            raw1 = cv2.cvtColor(img1_in, cv2.COLOR_RGB2GRAY) if len(img1_in.shape) == 3 else img1_in
            raw1 = cv2.resize(raw1, (240, 320))
            clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
            raw1 = clahe.apply(raw1)
        if img2_in is None:
            raw2 = cv2.imread("real_fingerprints/02_real_broken_scars.png", cv2.IMREAD_GRAYSCALE)
            raw2 = cv2.resize(raw2, (240, 320)) if raw2 is not None else generate_synthetic_fingerprint_advanced(pattern_type="loop", seed=99)
        else:
            raw2 = cv2.cvtColor(img2_in, cv2.COLOR_RGB2GRAY) if len(img2_in.shape) == 3 else img2_in
            raw2 = cv2.resize(raw2, (240, 320))
            clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
            raw2 = clahe.apply(raw2)
    else:
        # Kich ban A, B, C, D, E, F: Bien doi xoay goc tren cung 1 ngon tay
        raw1 = real1
        rot_angle = 0.0
        shift = (0, 0)
        if "B" in choice_str or "3°" in choice_str or "3 độ" in choice_str:
            rot_angle = 3.0
            shift = (3, -2)
        elif "C" in choice_str or "45°" in choice_str or "45 độ" in choice_str:
            rot_angle = 45.0
            shift = (-4, 4)
        elif "D" in choice_str or "90°" in choice_str or "90 độ" in choice_str:
            rot_angle = 90.0
            shift = (4, -4)
        elif "E" in choice_str or "180°" in choice_str or "180 độ" in choice_str or "lộn ngược" in choice_str.lower():
            rot_angle = 180.0
            shift = (2, 2)
        elif "F" in choice_str or "Tùy chỉnh" in choice_str or "Slider" in choice_str:
            rot_angle = float(custom_rot_param)
            shift = (2, -3)

        h, w = raw1.shape
        M = cv2.getRotationMatrix2D((w / 2.0, h / 2.0), rot_angle, 1.0)
        M[0, 2] += shift[0]
        M[1, 2] += shift[1]
        raw2 = cv2.warpAffine(raw1, M, (w, h), borderMode=cv2.BORDER_REFLECT)
        noise = np.random.normal(0, 3.0, (h, w)).astype(np.float32)
        raw2 = np.clip(raw2.astype(np.float32) + noise, 0, 255).astype(np.uint8)

    m1, me1, qc1 = segment_roi_with_qc(raw1)
    if qc1["is_valid"]:
        n1 = normalize_roi(raw1, m1)
        o1, _ = estimate_orientation_ridge(n1, m1)
        l1 = estimate_ridge_wavelength(n1, m1)
        e1 = apply_gabor_filter_bank(n1, o1, m1, lambd=l1)
        _, b1 = cv2.threshold(e1, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
        b1[m1 == 0] = 0
        sk1 = smart_prune_spurs(fast_thinning_zhang_suen(b1, m1))
        end1, bif1 = extract_validated_minutiae(sk1, o1, me1)
        pts1 = end1 + bif1
    else:
        pts1, end1, bif1 = [], [], []

    m2, me2, qc2 = segment_roi_with_qc(raw2)
    if qc2["is_valid"]:
        n2 = normalize_roi(raw2, m2)
        o2, _ = estimate_orientation_ridge(n2, m2)
        l2 = estimate_ridge_wavelength(n2, m2)
        e2 = apply_gabor_filter_bank(n2, o2, m2, lambd=l2)
        _, b2 = cv2.threshold(e2, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
        b2[m2 == 0] = 0
        sk2 = smart_prune_spurs(fast_thinning_zhang_suen(b2, m2))
        end2, bif2 = extract_validated_minutiae(sk2, o2, me2)
        pts2 = end2 + bif2
    else:
        pts2, end2, bif2 = [], [], []

    score, matched_pairs, align_info = match_minutiae_rigid_alignment(pts1, pts2, max_dist=max_dist_param, return_info=True)
    is_match = score >= threshold_param

    c1 = cv2.cvtColor(raw1, cv2.COLOR_GRAY2BGR)
    c2 = cv2.cvtColor(raw2, cv2.COLOR_GRAY2BGR)
    h, w = raw1.shape
    match_canvas = np.hstack((c1, c2))
    w_offset = w

    for p1, p2 in matched_pairs:
        p2_offset = (p2[0] + w_offset, p2[1])
        cv2.line(match_canvas, p1, p2_offset, (0, 229, 255), 1, cv2.LINE_AA)
        cv2.circle(match_canvas, p1, 3, (0, 0, 255), -1)
        cv2.circle(match_canvas, p2_offset, 3, (0, 255, 128), -1)

    banner_color = (0, 160, 60) if is_match else (0, 0, 200)
    cv2.rectangle(match_canvas, (10, 10), (w * 2 - 10, 48), banner_color, -1)
    status_label = "MATCH (AUTHENTICATED - GENUINE)" if is_match else "REJECTED (IMPOSTOR / NON-MATCH)"
    cv2.putText(match_canvas, f"SCORE: {score:.1f}% | ROT: {align_info['rot_deg']:+.1f} deg | {status_label}", (20, 36),
                cv2.FONT_HERSHEY_SIMPLEX, 0.58, (255, 255, 255), 2, cv2.LINE_AA)

    out_canvas = cv2.cvtColor(match_canvas, cv2.COLOR_BGR2RGB)
    match_time_ms = (time.time() - t0) * 1000.0

    verdict_str = "AUTHENTICATED (GENUINE MATCH - PASS)" if is_match else "REJECTED (IMPOSTOR / NON-MATCH)"
    report_md = f"""
### KẾT QUẢ ĐỐI SOÁT & XÁC THỰC SINH TRẮC HỌC (BIOMETRIC VERIFICATION V5)
- **Điểm đối sánh (Matching Score):** **`{score:.2f}%`** (Ngưỡng chấp thuận hoạt động: `{threshold_param}%`)
- **Phán quyết xác thực:** **`{verdict_str}`**
- **Góc xoay ước lượng (Estimated Rotation):** **`{align_info['rot_deg']:+.1f}°`** (Thuật toán Coarse-to-Fine 3D Hough Voting bất biến toàn dải 360°)
- **Độ dời không gian (Translation):** (Δx = {align_info['dx']:+.1f} px, Δy = {align_info['dy']:+.1f} px)
- **Số cặp Minutiae tương thích:** **`{len(matched_pairs)}` cặp điểm** (Ràng buộc kép: Cùng loại, |Δθ| ≤ 30°, bán kính đàn hồi phi tuyến)
- **Mẫu lưu trữ F1 (Template):** `{len(pts1)}` điểm Minutiae (`{len(end1)}` Kết thúc, `{len(bif1)}` Rẽ nhánh) - {qc1['status_text']}
- **Mẫu kiểm tra F2 (Query):** `{len(pts2)}` điểm Minutiae (`{len(end2)}` Kết thúc, `{len(bif2)}` Rẽ nhánh) - {qc2['status_text']}
- **Thời gian thực thi đối soát:** `{match_time_ms:.1f} ms` (Mảng tích lũy Accumulator tối ưu, đáp ứng thời gian thực < 50ms)
"""
    return out_canvas, report_md


def run_ablation_experiment(input_img, use_gabor, use_spurious_filter, num_orientations, min_dist_val):
    """Tab 3: Thuc nghiem triet tieu cac module tien xu ly va trich xuat"""
    t0 = time.time()
    if input_img is None:
        gray = cv2.imread("real_fingerprints/01_real_normal_clear.png", cv2.IMREAD_GRAYSCALE)
        if gray is None:
            gray = generate_synthetic_fingerprint_advanced(pattern_type="whorl", seed=42)
    else:
        gray = cv2.cvtColor(input_img, cv2.COLOR_RGB2GRAY) if len(input_img.shape) == 3 else input_img
        gray = cv2.resize(gray, (240, 320))
        clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
        gray = clahe.apply(gray)

    gray = cv2.resize(gray, (240, 320))
    mask, mask_eroded, qc_info = segment_roi_with_qc(gray)
    norm = normalize_roi(gray, mask)
    orient_ridge, _ = estimate_orientation_ridge(norm, mask)

    # 1. Ablation Gabor Filter
    if use_gabor:
        lambd_est = estimate_ridge_wavelength(norm, mask)
        enhanced = apply_gabor_filter_bank(norm, orient_ridge, mask, num_orientations=int(num_orientations), lambd=lambd_est)
    else:
        enhanced = norm.copy()

    # Nhị phân hóa
    _, binary = cv2.threshold(enhanced, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
    binary[mask == 0] = 0

    # Làm mảnh Skeleton
    skel = fast_thinning_zhang_suen(binary, mask)

    # 2. Ablation Spurious Filtering & Spur Pruning
    if use_spurious_filter:
        skel = smart_prune_spurs(skel, max_spur_length=6)
        endings, bifurcations = extract_validated_minutiae(skel, orient_ridge, mask_eroded, min_dist=min_dist_val)
    else:
        endings = []
        bifurcations = []
        pad = np.pad(skel, 1, mode='constant', constant_values=0)
        h, w = skel.shape
        for y in range(h):
            for x in range(w):
                if skel[y, x] > 0 and mask[y, x] > 0:
                    p = pad[y:y+3, x:x+3].flatten()
                    neighbors = [p[1], p[2], p[5], p[8], p[7], p[6], p[3], p[0]]
                    cn = sum(neighbors[k] == 0 and neighbors[(k + 1) % 8] > 0 for k in range(8))
                    if cn == 1:
                        endings.append({"x": x, "y": y, "type": "ending", "theta": 0.0})
                    elif cn == 3:
                        bifurcations.append({"x": x, "y": y, "type": "bifurcation", "theta": 0.0})

    vis_min = visualize_minutiae_map(skel, endings, bifurcations)
    total_time_ms = (time.time() - t0) * 1000.0

    report_md = f"""
### KẾT QUẢ THỰC NGHIỆM TRIỆT TIÊU (ABLATION AUDIT)
- **Cấu hình kiểm thử:**
  - Ngân hàng lọc Gabor 2D: **`{'BẬT (' + str(num_orientations) + ' hướng)' if use_gabor else 'TẮT (Dùng trực tiếp ảnh Normalized)'}`**
  - Bộ lọc Spurious & Khử nhánh gai: **`{'BẬT (Bán kính min ' + str(min_dist_val) + 'px)' if use_spurious_filter else 'TẮT (Giữ nguyên toàn bộ CN thô)'}`**
- **Số điểm trích xuất:**
  - Điểm kết thúc dải vân (Đỏ): **`{len(endings)}` điểm**
  - Điểm rẽ nhánh dải vân (Xanh): **`{len(bifurcations)}` điểm**
  - Tổng số Minutiae ghi nhận: **`{len(endings) + len(bifurcations)}` điểm**
- **Nhận định học thuật:**
  - {'Khi tắt Gabor: Ảnh nhị phân xuất hiện các vùng đứt đoạn loang lổ, sinh ra minutiae giả mật độ cao.' if not use_gabor else 'Bộ lọc Gabor tái tạo dòng vân mượt mà, triệt tiêu nhiễu đốm và nối liền vạch đứt.'}
  - {'Khi tắt bộ lọc Spurious: Các cặp Minutiae giả xuất hiện dày đặc quanh mép biên ROI và các vết gai rác.' if not use_spurious_filter else 'Bộ lọc Spurious đã loại trừ các cặp đối xứng giả và co biên an toàn.'}
- **Thời gian thực thi:** `{total_time_ms:.1f} ms`
"""
    return cv2.cvtColor(vis_min, cv2.COLOR_BGR2RGB), cv2.cvtColor(skel, cv2.COLOR_GRAY2RGB), report_md
```
